```python
# =========================================================
# EPU DAILY AND MONTHLY - RESEARCH PERIOD VALIDATION
# =========================================================

import pandas as pd

# Load actual EPU observations
epu_daily = pd.read_csv(
    RAW_DIR / "epu_daily_raw.csv",
    parse_dates=["observation_date"]
)

epu_monthly = pd.read_csv(
    RAW_DIR / "epu_monthly_raw.csv",
    parse_dates=["observation_date"]
)

# Select research period
start_date = "2010-01-01"
end_date = "2025-12-31"

epu_daily_period = epu_daily.loc[
    epu_daily["observation_date"].between(start_date, end_date)
].copy()

epu_monthly_period = epu_monthly.loc[
    epu_monthly["observation_date"].between(start_date, end_date)
].copy()

# Validate each dataset
for name, df, value_column in [
    ("EPU DAILY", epu_daily_period, "USEPUINDXD"),
    ("EPU MONTHLY", epu_monthly_period, "USEPUINDXM")
]:

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Observations:", len(df))

    print("First date:", df["observation_date"].min())
    print("Last date:", df["observation_date"].max())

    print("Missing values:")
    print(df.isna().sum())

    print("Duplicate dates:")
    print(df["observation_date"].duplicated().sum())

    print("\nDescriptive statistics:")
    display(df[value_column].describe())

    print("\nFirst 5 rows:")
    display(df.head())

    print("\nLast 5 rows:")
    display(df.tail())
```

```python
# =========================================================
# FINAL GPR VALIDATION
# Research period: 2010–2025
# =========================================================

import pandas as pd

START_DATE = "2010-01-01"
END_DATE = "2025-12-31"

# Ensure date columns are datetime
gpr_daily["date"] = pd.to_datetime(gpr_daily["date"])
gpr_monthly["month"] = pd.to_datetime(gpr_monthly["month"])

# Filter research period
gpr_daily_period = gpr_daily[
    gpr_daily["date"].between(START_DATE, END_DATE)
].copy()

gpr_monthly_period = gpr_monthly[
    gpr_monthly["month"].between(START_DATE, END_DATE)
].copy()

# Validate daily GPR
print("=" * 60)
print("GPR DAILY VALIDATION")
print("=" * 60)

print("Observations:", len(gpr_daily_period))
print("First date:", gpr_daily_period["date"].min())
print("Last date:", gpr_daily_period["date"].max())

print("\nMissing GPR values:")
print(gpr_daily_period["GPRD"].isna().sum())

print("\nDuplicate dates:")
print(gpr_daily_period["date"].duplicated().sum())

print("\nDescriptive statistics:")
display(gpr_daily_period["GPRD"].describe())

# Validate monthly GPR
print("\n" + "=" * 60)
print("GPR MONTHLY VALIDATION")
print("=" * 60)

print("Observations:", len(gpr_monthly_period))
print("First date:", gpr_monthly_period["month"].min())
print("Last date:", gpr_monthly_period["month"].max())

print("\nMissing GPR values:")
print(gpr_monthly_period["GPR"].isna().sum())

print("\nDuplicate dates:")
print(gpr_monthly_period["month"].duplicated().sum())

print("\nDescriptive statistics:")
display(gpr_monthly_period["GPR"].describe())
```

```python
# =========================================================
# GPR DATE COMPLETENESS CHECK
# =========================================================

expected_daily = pd.date_range(
    "2010-01-01",
    "2025-12-31",
    freq="D"
)

expected_monthly = pd.date_range(
    "2010-01-01",
    "2025-12-01",
    freq="MS"
)

missing_gpr_daily = expected_daily.difference(
    gpr_daily_period["date"]
)

missing_gpr_monthly = expected_monthly.difference(
    gpr_monthly_period["month"]
)

print("Missing GPR daily dates:", len(missing_gpr_daily))

print("Missing GPR monthly dates:", len(missing_gpr_monthly))

print("\nFirst 10 missing daily dates:")
print(missing_gpr_daily[:10])
```
